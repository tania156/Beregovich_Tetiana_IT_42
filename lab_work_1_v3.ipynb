```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
print(a)
print("shape:", a.shape)
print("dtype:", a.dtype)
```

```python
temps = np.array([-1, 0, 4, 10, 16, 20, 23, 22, 17, 11, 5, 1])
days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])
month_numbers = np.arange(1, 13)

print("temps shape:", temps.shape, "dtype:", temps.dtype)
print("days_in_month shape:", days_in_month.shape, "dtype:", days_in_month.dtype)
print("month_numbers shape:", month_numbers.shape, "dtype:", month_numbers.dtype)
```

```python
temps_fahrenheit = temps * 9 / 5 + 32
print("Температура у °F:", temps_fahrenheit)

warm_days_total = days_in_month[temps > 15].sum()
print("Кількість днів з T > 15°C:", warm_days_total)

print("Сума днів у році:", days_in_month.sum())
```

```python
print(f"Найтепліший місяць: №{temps.argmax() + 1} ({temps.max()}°C)")
print(f"Найхолодніший місяць: №{temps.argmin() + 1} ({temps.min()}°C)")
print("Місяці з T < 0°C:", month_numbers[temps < 0])
print("Відсортовані температури:", np.sort(temps))
```

```python
sun_hours = np.array([60, 80, 140, 180, 260, 300, 330, 300, 230, 160, 80, 50])
data_2d = np.array([temps, days_in_month, sun_hours])

print("Суми по рядках:", data_2d.sum(axis=1))
print("Середні по рядках:", data_2d.mean(axis=1))

col_sums = data_2d.sum(axis=0)
print("Суми по стовпцях:", col_sums)
print("Сума для Січня (1-й стовпець):", col_sums[0])
```

```python
rows = np.array([temps, days_in_month])
correction = np.array([1, 0])

try:
    print(rows + correction)
except ValueError as e:
    print("Очікувана помилка:", e)

correction_reshaped = correction.reshape(2, 1)
print("Результат після reshape:\n", rows + correction_reshaped)
```

```python
## Відповіді на питання

1. **Різниця між (12,) та (1, 12):** (12,) — це 1D вектор з 12 елементів, а (1, 12) — це 2D матриця з 1 рядком та 12 стовпцями.
2. **Довжина temps[temps > 15]:** Маска обирає тільки ті елементи, де умова True, тому залишаються лише місяці, де температура дійсно вища за 15°C.
3. **Додавання (3,) до (3, 12):** Виникне помилка `ValueError`, оскільки останні розмірності (3 та 12) не збігаються і жодна не дорівнює 1.
4. **Чому reshape(2, 1) працює:** Спочатку порівнювалися 12 і 2 (несумісні). Після `.reshape(2, 1)` порівнюються 12 і 1 — осі з розміром 1 розтягуються автоматично за правилами broadcasting.
```
